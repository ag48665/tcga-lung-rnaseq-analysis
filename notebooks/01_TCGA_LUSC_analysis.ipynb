```python
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind

tumor_samples = labels[labels == "tumor"].index
normal_samples = labels[labels == "normal"].index

tumor = expr[tumor_samples]
normal = expr[normal_samples]

log2fc = np.log2(tumor.mean(axis=1) + 1) - np.log2(normal.mean(axis=1) + 1)

tstat, pvals = ttest_ind(
    np.log1p(tumor).T.values,    
    np.log1p(normal).T.values,    
    axis=0,
    equal_var=False,
    nan_policy="omit"
)

deg = pd.DataFrame({
    "log2FC": log2fc.values,
    "pvalue": pvals
}, index=expr.index)

deg["-log10p"] = -np.log10(deg["pvalue"] + 1e-300)

deg_sorted = deg.sort_values("pvalue")
deg_sorted.head(20)
```
